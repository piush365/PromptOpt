```python
import pandas as pd
from datasets import load_dataset

print("Loading yahma/alpaca-cleaned dataset...")
dataset = load_dataset("yahma/alpaca-cleaned", split="train")
df = dataset.to_pandas()

# Filter out rows that have extra 'input' data. 
# We only want standalone instructions for prompt compression.
df_standalone = df[df['input'] == ""]
print(f"Found {len(df_standalone)} standalone instructions.")

# Sample 1,000 rows for your LoRA
SAMPLE_SIZE = 1000
sample_df = df_standalone.sample(n=SAMPLE_SIZE, random_state=42)

# Prepare the dataframe for your manual annotation
annotation_df = pd.DataFrame()
annotation_df['optimized_target'] = sample_df['instruction'] # The short Alpaca prompt
annotation_df['unoptimized_rambling_prompt'] = ""            # Your blank column for fluff

# Export to CSV
csv_filename = "alpaca_for_annotation.csv"
annotation_df.to_csv(csv_filename, index=False)

print(f"\nSuccess! Exported {SAMPLE_SIZE} rows to '{csv_filename}'.")
print("Download this file and start adding conversational fluff to the blank column.")
```
